```python
# Calculate duration directly from the provided timestamps

flights["calculated_duration_raw"] = (
    flights["arrival_time"] - flights["departure_time"]
)

print(
    flights[
        ["flight_id", "departure_time", "arrival_time",
         "duration", "calculated_duration_raw"]
    ].head(20).to_string(index=False)
)
```

```python
# Convert duration values safely into timedeltas

def convert_duration(value):
    if pd.isna(value):
        return pd.NaT
    
    if isinstance(value, pd.Timestamp):
        return pd.Timedelta(
            hours=value.hour,
            minutes=value.minute,
            seconds=value.second,
            microseconds=value.microsecond
        )
    
    if hasattr(value, "hour") and hasattr(value, "minute") and hasattr(value, "second"):
        return pd.Timedelta(
            hours=value.hour,
            minutes=value.minute,
            seconds=value.second,
            microseconds=getattr(value, "microsecond", 0)
        )
    
    return pd.to_timedelta(str(value), errors="coerce")


provided_duration = flights["duration"].apply(convert_duration)

duration_difference = (
    provided_duration - flights["calculated_duration_raw"]
)

print("Records with duration mismatch:",
      (duration_difference != pd.Timedelta(0)).sum())

print("\nLargest mismatches:")
print(
    flights.loc[
        duration_difference != pd.Timedelta(0),
        ["flight_id", "departure_time", "arrival_time",
         "duration", "calculated_duration_raw"]
    ].head(20).to_string(index=False)
)
```

```python
from datetime import datetime

today = pd.Timestamp("2026-09-10")

calculated_age = (
    today.year
    - passengers["date_of_birth"].dt.year
    - (
        (today.month < passengers["date_of_birth"].dt.month)
        |
        (
            (today.month == passengers["date_of_birth"].dt.month)
            & (today.day < passengers["date_of_birth"].dt.day)
        )
    )
)

age_mismatch = passengers[calculated_age != passengers["age"]].copy()

print("Age/date-of-birth mismatches:", len(age_mismatch))

print("\nSample mismatches:")
print(
    age_mismatch[
        ["passenger_id", "age", "date_of_birth"]
    ].head(20).to_string(index=False)
)
```

```python
calculated_age_booking = (
    bookings[["passenger_id", "booking_date"]]
    .merge(
        passengers[["passenger_id", "age", "date_of_birth"]],
        on="passenger_id",
        how="left"
    )
)

calculated_age_booking["age_at_booking"] = (
    calculated_age_booking["booking_date"].dt.year
    - calculated_age_booking["date_of_birth"].dt.year
    - (
        (calculated_age_booking["booking_date"].dt.month < calculated_age_booking["date_of_birth"].dt.month)
        |
        (
            (calculated_age_booking["booking_date"].dt.month == calculated_age_booking["date_of_birth"].dt.month)
            & (calculated_age_booking["booking_date"].dt.day < calculated_age_booking["date_of_birth"].dt.day)
        )
    )
)

mismatch_booking = calculated_age_booking[
    calculated_age_booking["age"] != calculated_age_booking["age_at_booking"]
]

print("Total booking-passenger records:", len(calculated_age_booking))
print("Age mismatches at booking date:", len(mismatch_booking))
print("Matching age at booking date:",
      len(calculated_age_booking) - len(mismatch_booking))
```

```python
# STEP 12: Identify malformed flight IDs

import re

flight_id_pattern = r"^[A-Z0-9]{2}[0-9]{3}$"

malformed_flights = flights[
    ~flights["flight_id"].astype(str).str.match(
        flight_id_pattern,
        na=False
    )
]

print("Flight ID validation completed!")

print("\nTotal flight records:", len(flights))

print("Valid flight IDs:",
      len(flights) - len(malformed_flights))

print("Malformed flight IDs:",
      len(malformed_flights))

print("\nMalformed flight ID values:")

print(
    malformed_flights[
        ["flight_id", "airline", "source", "destination"]
    ].to_string(index=False)
)
```

```python
# STEP 13: Check flight ID frequency

flight_id_counts = flights["flight_id"].value_counts()

repeated_flight_ids = flight_id_counts[
    flight_id_counts > 1
]

print("Flight ID frequency check completed!")

print("\nUnique flight IDs:",
      flights["flight_id"].nunique())

print("Flight IDs appearing more than once:",
      len(repeated_flight_ids))

print("\nRepeated flight IDs:")

print(repeated_flight_ids.head(20))
```

```python
# STEP 11: Create flight analytics dataset

flight_analytics = flights.copy()

# Convert duration from time format to total minutes
flight_analytics["duration_minutes"] = (
    flight_analytics["duration"].apply(
        lambda x: x.hour * 60 + x.minute + x.second / 60
    )
)

# Create route
flight_analytics["route"] = (
    flight_analytics["source"]
    + " → "
    + flight_analytics["destination"]
)

# Extract date and time components
flight_analytics["departure_date"] = (
    flight_analytics["departure_time"].dt.date
)

flight_analytics["arrival_date"] = (
    flight_analytics["arrival_time"].dt.date
)

flight_analytics["departure_hour"] = (
    flight_analytics["departure_time"].dt.hour
)

flight_analytics["arrival_hour"] = (
    flight_analytics["arrival_time"].dt.hour
)

# Identify overnight flights
flight_analytics["is_overnight"] = (
    flight_analytics["arrival_date"]
    != flight_analytics["departure_date"]
)

print("Flight analytics dataset created successfully!")

print("\nShape:", flight_analytics.shape)

print("\nColumns:")
print(flight_analytics.columns.tolist())

print("\nOvernight flight distribution:")
print(
    flight_analytics["is_overnight"].value_counts()
)

print("\nSample:")
print(
    flight_analytics[
        [
            "flight_id",
            "airline",
            "source",
            "destination",
            "route",
            "duration_minutes",
            "departure_date",
            "arrival_date",
            "is_overnight"
        ]
    ].head(10)
)
```

```python
# STEP 12: Combine flight and booking analytics

final_analytics = booking_analytics.merge(
    flight_analytics,
    on="flight_id",
    how="left",
    suffixes=("_booking", "_flight")
)

print("Integrated analytics dataset created successfully!")

print("\nShape:", final_analytics.shape)

print("\nColumns:")
print(final_analytics.columns.tolist())

print("\nMissing flight matches:",
      final_analytics["airline"].isna().sum())

print("\nSample:")
print(
    final_analytics[
        [
            "booking_id",
            "flight_id",
            "airline",
            "source",
            "destination",
            "route",
            "status",
            "payment_count",
            "valid_payment_amount",
            "duration_minutes",
            "is_overnight",
            "payment_status_issue"
        ]
    ].head(10)
)
```

```python
# STEP 13: Identify duplicated booking records after flight merge

duplicated_bookings = final_analytics[
    final_analytics.duplicated(
        subset="booking_id",
        keep=False
    )
].sort_values("booking_id")

print("Duplicated bookings after flight merge:",
      duplicated_bookings["booking_id"].nunique())

print("\nTotal affected rows:",
      len(duplicated_bookings))

print("\nAffected records:")

print(
    duplicated_bookings[
        [
            "booking_id",
            "flight_id",
            "airline",
            "source",
            "destination",
            "departure_time",
            "arrival_time"
        ]
    ].to_string(index=False)
)
```

```python
# STEP 13: Create a safe flight lookup without ambiguous flight IDs

# Identify flight IDs that occur more than once
ambiguous_flight_ids = (
    flights["flight_id"]
    .value_counts()
    .loc[lambda x: x > 1]
    .index
)

# Keep only flight IDs that uniquely identify one flight record
flight_lookup = flight_analytics[
    ~flight_analytics["flight_id"].isin(ambiguous_flight_ids)
].copy()

print("Safe flight lookup created!")

print("\nOriginal flight records:", len(flight_analytics))

print("Ambiguous flight IDs:",
      len(ambiguous_flight_ids))

print("Safe flight records:",
      len(flight_lookup))

print("\nAmbiguous flight IDs:")
print(ambiguous_flight_ids.tolist())
```

```python
# STEP 14: Create corrected integrated analytics dataset

final_analytics = booking_analytics.merge(
    flight_lookup,
    on="flight_id",
    how="left",
    suffixes=("_booking", "_flight")
)

print("Corrected integrated analytics dataset created successfully!")

print("\nShape:", final_analytics.shape)

print("\nMissing flight matches:",
      final_analytics["airline"].isna().sum())

print("\nBookings retained:",
      final_analytics["booking_id"].nunique())

print("\nDuplicate booking IDs:",
      final_analytics["booking_id"].duplicated().sum())

print("\nBookings with ambiguous flight IDs:",
      final_analytics["flight_id"].isin(ambiguous_flight_ids).sum())

print("\nSample:")
print(
    final_analytics[
        [
            "booking_id",
            "flight_id",
            "airline",
            "source",
            "destination",
            "route",
            "status",
            "duration_minutes",
            "is_overnight",
            "payment_status_issue"
        ]
    ].head(10)
)
```

```python
# STEP 15: Flag flight matching/data-quality issues

final_analytics["flight_match_status"] = "Matched"

final_analytics.loc[
    final_analytics["flight_id"].isin(ambiguous_flight_ids),
    "flight_match_status"
] = "Ambiguous Flight ID"

final_analytics.loc[
    final_analytics["airline"].isna(),
    "flight_match_status"
] = "No Reliable Flight Match"

print("Flight match status created successfully!")

print("\nFlight match status:")
print(final_analytics["flight_match_status"].value_counts())

print("\nAffected bookings:")
print(
    final_analytics[
        final_analytics["flight_match_status"] != "Matched"
    ][
        [
            "booking_id",
            "flight_id",
            "flight_match_status"
        ]
    ]
)
```

```python
# STEP 16: Create privacy-protected analytical dataset

import hashlib

protected_analytics = final_analytics.copy()

# Hash passport numbers
protected_analytics["passport_hash"] = (
    protected_analytics["passport_number"]
    .astype(str)
    .apply(lambda x: hashlib.sha256(x.encode()).hexdigest())
)

# Mask emergency contact names
protected_analytics["emergency_contact_name_masked"] = (
    protected_analytics["emergency_contact_name"]
    .astype(str)
    .apply(
        lambda x: (
            x[0] + "***" + x[-1]
            if len(x) > 2 and x != "nan"
            else "***"
        )
    )
)

# Mask emergency contact phone numbers
protected_analytics["emergency_contact_phone_masked"] = (
    protected_analytics["emergency_contact_phone"]
    .astype(str)
    .apply(
        lambda x: (
            "*" * max(0, len(x) - 4) + x[-4:]
            if len(x) >= 4 and x != "nan"
            else "***"
        )
    )
)

# Remove raw PII from the analytical copy
protected_analytics = protected_analytics.drop(
    columns=[
        "passport_number",
        "emergency_contact_name",
        "emergency_contact_phone"
    ]
)

print("Privacy-protected analytical dataset created successfully!")

print("\nShape:", protected_analytics.shape)

print("\nRaw PII columns remaining:")
print([
    col for col in [
        "passport_number",
        "emergency_contact_name",
        "emergency_contact_phone"
    ]
    if col in protected_analytics.columns
])

print("\nProtected PII columns:")
print([
    col for col in [
        "passport_hash",
        "emergency_contact_name_masked",
        "emergency_contact_phone_masked"
    ]
    if col in protected_analytics.columns
])

print("\nSample:")
print(
    protected_analytics[
        [
            "booking_id",
            "passenger_id",
            "passport_hash",
            "emergency_contact_name_masked",
            "emergency_contact_phone_masked"
        ]
    ].head()
)
```

```python
# STEP 17: Create core flight KPIs

average_flight_duration = flight_analytics["duration_minutes"].mean()

total_flights = len(flight_analytics)

overnight_flights = flight_analytics["is_overnight"].sum()

overnight_percentage = (
    overnight_flights / total_flights
) * 100

airline_distribution = (
    flight_analytics["airline"]
    .value_counts()
    .reset_index()
)

airline_distribution.columns = [
    "airline",
    "flight_count"
]

route_traffic = (
    flight_analytics["route"]
    .value_counts()
    .reset_index()
)

route_traffic.columns = [
    "route",
    "flight_count"
]

print("Core flight KPIs created successfully!")

print("\nAverage flight duration (minutes):",
      round(average_flight_duration, 2))

print("Total flights:", total_flights)

print("Overnight flights:", overnight_flights)

print("Overnight flight percentage:",
      round(overnight_percentage, 2), "%")

print("\nAirline distribution:")
print(airline_distribution)

print("\nTop 10 routes by flight count:")
print(route_traffic.head(10))
```

```python
# STEP 18: Create booking and operational KPIs

total_bookings = len(protected_analytics)

status_distribution = (
    protected_analytics["status"]
    .value_counts()
    .reset_index()
)

status_distribution.columns = [
    "booking_status",
    "booking_count"
]

cancelled_bookings = (
    protected_analytics["status"] == "CANCELLED"
).sum()

cancellation_rate = (
    cancelled_bookings / total_bookings
) * 100

payment_issue_distribution = (
    protected_analytics["payment_status_issue"]
    .value_counts()
    .reset_index()
)

payment_issue_distribution.columns = [
    "payment_status_issue",
    "booking_count"
]

flight_match_distribution = (
    protected_analytics["flight_match_status"]
    .value_counts()
    .reset_index()
)

flight_match_distribution.columns = [
    "flight_match_status",
    "booking_count"
]

print("Booking and operational KPIs created successfully!")

print("\nTotal bookings:", total_bookings)

print("Cancelled bookings:", cancelled_bookings)

print("Cancellation rate:",
      round(cancellation_rate, 2), "%")

print("\nBooking status distribution:")
print(status_distribution)

print("\nPayment issue distribution:")
print(payment_issue_distribution)

print("\nFlight match status:")
print(flight_match_distribution)
```

```python
# STEP 19: Airline performance summary

airline_performance = (
    flight_analytics
    .groupby("airline")
    .agg(
        flight_count=("flight_id", "count"),
        average_duration_minutes=("duration_minutes", "mean"),
        overnight_flights=("is_overnight", "sum")
    )
    .reset_index()
)

airline_performance["overnight_percentage"] = (
    airline_performance["overnight_flights"]
    / airline_performance["flight_count"]
    * 100
)

airline_performance["average_duration_minutes"] = (
    airline_performance["average_duration_minutes"].round(2)
)

airline_performance["overnight_percentage"] = (
    airline_performance["overnight_percentage"].round(2)
)

airline_performance = airline_performance.sort_values(
    "flight_count",
    ascending=False
)

print("Airline performance summary created successfully!")

print("\nAirline performance:")
print(airline_performance)
```

```python
# STEP 20: Route performance summary

route_performance = (
    flight_analytics
    .groupby("route")
    .agg(
        flight_count=("flight_id", "count"),
        average_duration_minutes=("duration_minutes", "mean"),
        overnight_flights=("is_overnight", "sum")
    )
    .reset_index()
)

route_performance["average_duration_minutes"] = (
    route_performance["average_duration_minutes"].round(2)
)

route_performance["overnight_percentage"] = (
    route_performance["overnight_flights"]
    / route_performance["flight_count"]
    * 100
).round(2)

route_performance = route_performance.sort_values(
    "flight_count",
    ascending=False
)

print("Route performance summary created successfully!")

print("\nTotal routes:", len(route_performance))

print("\nTop 15 routes:")
print(route_performance.head(15))
```

```python
# STEP 21: Create operational anomaly summary

unknown_airline_flights = (
    flight_analytics["airline"] == "UNKNOWN"
).sum()

ambiguous_flight_bookings = (
    protected_analytics["flight_match_status"] == "No Reliable Flight Match"
).sum()

payment_issue_bookings = (
    protected_analytics["payment_status_issue"] != "No Issue"
).sum()

missing_payment_amount_bookings = (
    protected_analytics["payment_status_issue"] == "Missing Payment Amount"
).sum()

operational_anomaly_summary = pd.DataFrame({
    "anomaly_type": [
        "Unknown Airline",
        "Payment Status Issue",
        "Missing Payment Amount",
        "Ambiguous Flight ID"
    ],
    "record_count": [
        unknown_airline_flights,
        payment_issue_bookings,
        missing_payment_amount_bookings,
        ambiguous_flight_bookings
    ]
})

print("Operational anomaly summary created successfully!")

print("\nOperational anomalies:")
print(operational_anomaly_summary)

print("\nTotal bookings with payment-related issues:",
      payment_issue_bookings)

print("Bookings with ambiguous flight IDs:",
      ambiguous_flight_bookings)

print("Flights with UNKNOWN airline:",
      unknown_airline_flights)
```

```python
# STEP 22: Create final KPI summary

kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Flights",
        "Total Bookings",
        "Average Flight Duration (min)",
        "Overnight Flights",
        "Overnight Flight %",
        "Cancelled Bookings",
        "Cancellation Rate %",
        "Payment Issue Bookings",
        "Unknown Airline Flights",
        "Ambiguous Flight Bookings"
    ],
    "Value": [
        total_flights,
        total_bookings,
        round(average_flight_duration, 2),
        overnight_flights,
        round(overnight_percentage, 2),
        cancelled_bookings,
        round(cancellation_rate, 2),
        payment_issue_bookings,
        unknown_airline_flights,
        ambiguous_flight_bookings
    ]
})

print("Final KPI summary created successfully!")

print("\nkpi_summary:")
print(kpi_summary)
```
